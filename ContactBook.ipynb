```python
import json
import os

# ==========================================
#           CONTACT BOOK APPLICATION
# ==========================================

FILE_NAME = "contacts.json"


class ContactBook:

    def __init__(self):
        self.contacts = self.load_contacts()

    # Load contacts from file
    def load_contacts(self):
        if os.path.exists(FILE_NAME):
            try:
                with open(FILE_NAME, "r") as file:
                    return json.load(file)
            except (json.JSONDecodeError, OSError):
                return []
        return []

    # Save contacts to file
    def save_contacts(self):
        with open(FILE_NAME, "w") as file:
            json.dump(self.contacts, file, indent=4)

    # Add a new contact
    def add_contact(self):
        print("\n========== ADD CONTACT ==========")

        name = input("Enter name: ").strip()
        phone = input("Enter phone number: ").strip()
        email = input("Enter email: ").strip()
        address = input("Enter address: ").strip()

        if not name or not phone:
            print("Error: Name and phone number are required!")
            return

        # Check for duplicate phone numbers
        for contact in self.contacts:
            if contact["phone"] == phone:
                print("A contact with this phone number already exists!")
                return

        contact = {
            "name": name,
            "phone": phone,
            "email": email,
            "address": address
        }

        self.contacts.append(contact)
        self.save_contacts()

        print(f"\nContact '{name}' added successfully!")

    # View all contacts
    def view_contacts(self):
        print("\n========== CONTACT LIST ==========")

        if not self.contacts:
            print("No contacts found.")
            return

        for index, contact in enumerate(self.contacts, start=1):
            print(f"\nContact {index}")
            print(f"Name    : {contact['name']}")
            print(f"Phone   : {contact['phone']}")
            print(f"Email   : {contact['email']}")
            print(f"Address : {contact['address']}")
            print("-" * 35)

    # Search contacts
    def search_contact(self):
        print("\n========== SEARCH CONTACT ==========")

        query = input(
            "Enter name or phone number to search: "
        ).strip().lower()

        results = [
            contact for contact in self.contacts
            if query in contact["name"].lower()
            or query in contact["phone"].lower()
        ]

        if results:
            print(f"\nFound {len(results)} matching contact(s):")

            for contact in results:
                print(f"\nName    : {contact['name']}")
                print(f"Phone   : {contact['phone']}")
                print(f"Email   : {contact['email']}")
                print(f"Address : {contact['address']}")
        else:
            print("No matching contact found.")

    # Update a contact
    def update_contact(self):
        print("\n========== UPDATE CONTACT ==========")

        phone = input(
            "Enter the current phone number of the contact: "
        ).strip()

        contact = next(
            (c for c in self.contacts if c["phone"] == phone),
            None
        )

        if contact is None:
            print("Contact not found!")
            return

        print("\nLeave a field blank to keep its current value.")

        name = input(f"New name [{contact['name']}]: ").strip()
        new_phone = input(
            f"New phone [{contact['phone']}]: "
        ).strip()
        email = input(f"New email [{contact['email']}]: ").strip()
        address = input(
            f"New address [{contact['address']}]: "
        ).strip()

        if new_phone and any(
            c["phone"] == new_phone and c is not contact
            for c in self.contacts
        ):
            print("Another contact already uses this phone number!")
            return

        if name:
            contact["name"] = name
        if new_phone:
            contact["phone"] = new_phone
        if email:
            contact["email"] = email
        if address:
            contact["address"] = address

        self.save_contacts()
        print("\nContact updated successfully!")

    # Delete a contact
    def delete_contact(self):
        print("\n========== DELETE CONTACT ==========")

        phone = input(
            "Enter the phone number of the contact to delete: "
        ).strip()

        contact = next(
            (c for c in self.contacts if c["phone"] == phone),
            None
        )

        if contact is None:
            print("Contact not found!")
            return

        confirm = input(
            f"Are you sure you want to delete "
            f"{contact['name']}? (yes/no): "
        ).strip().lower()

        if confirm == "yes":
            self.contacts.remove(contact)
            self.save_contacts()
            print("Contact deleted successfully!")
        else:
            print("Deletion cancelled.")

    # Main menu
    def run(self):
        while True:
            print("\n" + "=" * 40)
            print("          CONTACT BOOK")
            print("=" * 40)
            print("1. Add Contact")
            print("2. View All Contacts")
            print("3. Search Contact")
            print("4. Update Contact")
            print("5. Delete Contact")
            print("6. Exit")
            print("=" * 40)

            choice = input("Enter your choice (1-6): ").strip()

            if choice == "1":
                self.add_contact()

            elif choice == "2":
                self.view_contacts()

            elif choice == "3":
                self.search_contact()

            elif choice == "4":
                self.update_contact()

            elif choice == "5":
                self.delete_contact()

            elif choice == "6":
                print("\nThank you for using Contact Book!")
                print("Goodbye!")
                break

            else:
                print("Invalid choice! Please enter a number from 1 to 6.")


# Run the application
if __name__ == "__main__":
    app = ContactBook()
    app.run()
```
